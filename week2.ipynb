# Week 2: Data Types, Operators, and Algorithmic Representation

We learn:
- Basic data types
- Operators
- How to represent algorithms using pseudocode and flowcharts


```python
# Integer
age = 18

# Float
height = 1.72

# String
name = "Valentinus"

# Boolean
is_student = True

print(age, height, name, is_student)
```

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** b)
print("Modulus:", a % b)
```

## Pseudocode Example: Checking Temperature

IF temperature > 30:
    PRINT "Hot day"
ELSE:
    PRINT "Comfortable day"


```python
temperature = 28

if temperature > 30:
    print("Hot day")
else:
    print("Comfortable day")
```

```python
# Flowchart-like decision
score = 82

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
else:
    print("Grade: C")
```
