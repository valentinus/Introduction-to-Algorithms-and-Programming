## Algorithms in Daily Life
An algorithm is a step-by-step procedure to solve a problem.
Below is an example of an algorithm for making tea.

```python
steps_for_tea = [
    "Boil water",
    "Add tea leaves or tea bag",
    "Wait for 3 minutes",
    "Pour into cup",
    "Add sugar or milk if desired"
]

for step in steps_for_tea:
    print(step)
```
