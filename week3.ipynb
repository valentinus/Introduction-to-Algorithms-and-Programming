# Week 3: Expressions, Control Flow, and Iterations

We learn:
- Conditional statements
- Loops (for, while)
- How computers repeat tasks


```python
weather = "rainy"

if weather == "sunny":
    print("Go for a walk.")
elif weather == "rainy":
    print("Bring an umbrella.")
else:
    print("Check the weather forecast.")
```

```python
cities = ["Tokyo", "Kyoto", "Osaka", "Sapporo"]

for city in cities:
    print("Exploring:", city)
```

```python
count = 1

while count <= 5:
    print("Count:", count)
    count += 1
```

```python
sentence = "Our students come from many countries around the world"
words = sentence.split()

print("Number of words:", len(words))
```

```python
raw_data = [" Japan ", "  USA", "France ", "  Indonesia  "]

cleaned = []

for item in raw_data:
    cleaned.append(item.strip())

print(cleaned)
```
