# Week 3: Conditional Statements
In this notebook, we will learn:
- Flow of control
- Conditional statements
- One-way selection (if)
- Two-way selection (if-else)
- Multi-way selection (if-elif-else)
- Common mistakes

Example 1: Simple Condition

```python
temperature = 28

if temperature > 25:
    print("Hot")
```

Example 2: One-Way Selection

```python
raining = True

if raining:
    print("Take an umbrella")

print("Walk to class")
```

Example 3: Two-Way Selection

```python
temperature = 28

if temperature > 25:
    print("Hot")
else:
    print("Cool")

print("Have a good day!")
```

Example 4: Multiple Conditions

```python
score = 72
attendance = 9

if score >= 60 and attendance >= 9:
    print("You passed!")
    print("Well done")

print("Next class")
```

Example 5: Multi-Way Selection

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
```

Example 6: Museum Ticket Pricing

```python
age = 20
is_student = True

if age < 6:
    price = 0
elif age >= 65 or is_student:
    price = 500
else:
    price = 1000

print("Ticket price:", price)
```

#Common Mistakes
Mistake 1: Wrong Order of Conditions

```python
age = 5
is_student = True

if age >= 65 or is_student:
    price = 500
elif age < 6:
    price = 0
else:
    price = 1000

print(price)
```

Mistake 2: Forgetting to indent

```python
score = 72
if score >= 60:
    print("Well done")
print("You passed")
```

Mistake 3: Assigning instead of equality checking

```python
myBoolean = True
if MyBoolean = True:
    print("Success")

```
